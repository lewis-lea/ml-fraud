# Exploring data sources and discovering features

```python
%load_ext autoreload
%autoreload 2

import pandas
import duckdb
import ml_fraud
from ml_fraud.database import example_database as egdb
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Connect to the example database, building it if it does not exist
db = egdb.setup_example_database()
```

Access data from database

To begin, just get all of the features

```python
accounts = db.sql("SELECT * FROM accounts").df()
transactions = db.sql("SELECT * FROM transactions").df()
```

```python
# Visualise the account data items and their relationships
sns.pairplot(accounts,
             vars = ['txn_per_month_min', 
                     'txn_per_month_max',
                     'typical_amount',
                     'foreign_txn_prob',
                     'known_beneficiary_count'],
                     hue = 'persona',
                      diag_kws={"common_norm": False})
plt.show()
```

```python
# Visualise the transaction data items and their relationships
sns.pairplot(transactions,
             vars = ['amount', 
                     'time_since_last_txn', 
                     'hour_of_day',
                     'day_of_week'],
             hue = 'is_fraud',
             diag_kws={"common_norm": False})
plt.show()
```
